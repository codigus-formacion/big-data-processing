# Ejemplos de SQL en Databricks SQL Warehouse

En este notebook aprenderemos cómo crear algunos datos de prueba y realizar consultas sencillas sobre el SQL Warehouse de Databricks.

### 1. Creación de un schema y creación una tabla de ejemplo

Lo primero será crear un schema **liga** dentro de nuestro workspace.

```python
%sql
CREATE SCHEMA IF NOT EXISTS workspace.liga;
```

Tras esto, crearemos una tabla **equipos** con algunos datos de ejemplo. Para ello, ejecutaremos la siguiente consulta SQL:

```python
%sql
CREATE OR REPLACE TABLE workspace.liga.partidos (
  jornada          INT,
  equipo_local     STRING,
  equipo_visitante STRING,
  goles_local      INT,
  goles_visitante  INT
);
```

A a continuación, añadimos algunos datos de prueba

```python
%sql
INSERT INTO workspace.liga.partidos VALUES
  (1, 'Rayo Vallecano', 'Getafe',          2, 1),
  (1, 'Atlético',       'Real Madrid',     1, 1),
  (1, 'Barcelona',      'Valencia',        3, 0),
  (2, 'Getafe',         'Atlético',        0, 2),
  (2, 'Real Madrid',    'Barcelona',       2, 2),
  (2, 'Valencia',       'Rayo Vallecano',  1, 3);
```

### 02. Consultas básicas 

A continuación, podemos realizar algunas consultas SQL sobre la tabla **partidos** que acabamos de crear.

```python
%sql
-- Todos los partidos
SELECT * FROM workspace.liga.partidos;
```

```python
%sql
-- Goles por jornada
SELECT jornada, SUM(goles_local + goles_visitante) AS goles_totales
FROM workspace.liga.partidos
GROUP BY jornada
ORDER BY jornada;
```

```python
%sql
-- Partidos con más de tres goles
SELECT equipo_local, equipo_visitante, goles_local + goles_visitante AS goles
FROM workspace.liga.partidos
WHERE goles_local + goles_visitante > 3;
```

### 03. Comprobar historial de operaciones (Delta)

Podemos comprobar los cambios que se han hecho en una tabla

```python
%sql
-- "Nos hemos equivocado al teclear un resultado"
UPDATE workspace.liga.partidos
SET goles_local = 5
WHERE equipo_local = 'Barcelona';
```

```python
%sql
-- El historial de la tabla: cada operación, con su autor y su hora
DESCRIBE HISTORY workspace.liga.partidos;
```

```python
%sql
-- Consultar la tabla como estaba antes del error
SELECT * FROM workspace.liga.partidos VERSION AS OF 1;
```
